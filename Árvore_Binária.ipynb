Implementar uma estrutura de dados em árvore em Python.

·       Nó (com os atributos chave, filho esquerdo e filho direito), e

·       Árvore (com o atributo raíz (que tem o tipo nó))

A árvore deve ter os seguintes métodos:

·       Imprimir o elemento de máxima prioridade em uma árvore max-heap

·       Imprimir os elementos de uma árvore max-heap em um vetor

·       Inserir nó em uma árvore max-heap

·       Extrair o máximo em uma árvore max-heap

·       Inserir nó em uma árvore binária de busca

·       Remover um nó em uma árvore binária de busca

·       Imprimir os elementos de uma árvore binária de busca em um vetor segundo cada uma das estratégias pré-ordem, em ordem e pós-ordem.

·       Verificar se a árvore é max-heap

·       Verificar se a árvore é binária de busca

**DEFINIÇÕES DE CLASSES E MÉTODOS**

```python
class No:
    def __init__(self, chave):
        self.chave = chave
        self.esquerdo = None
        self.direito = None

class Arvore:
    def __init__(self):
        self.root = None

    def get_max(self):
        if self.root:
            return self.root.chave
        return None

    def max_heap_to_vector(self):
        if not self.root:
            return []
        resultado = []
        fila = [self.root]
        while fila:
            no = fila.pop(0)
            resultado.append(no.chave)
            if no.esquerdo:
                fila.append(no.esquerdo)
            if no.direito:
                fila.append(no.direito)
        return resultado

    def insert_max_heap(self, chave):
        if not self.root:
            self.root = No(chave)
            return

        fila = [self.root]
        while fila:
            no = fila.pop(0)
            if not no.esquerdo:
                no.esquerdo = No(chave)
                self._bubble_up(no.esquerdo)
                return
            else:
                fila.append(no.esquerdo)

            if not no.direito:
                no.direito = No(chave)
                self._bubble_up(no.direito)
                return
            else:
                fila.append(no.direito)

    def _bubble_up(self, node):
        parent = self._find_parent(self.root, node)
        while parent and node.chave > parent.chave:
            node.chave, parent.chave = parent.chave, node.chave
            node = parent
            parent = self._find_parent(self.root, node)

    def _find_parent(self, root, node):
        """Encontra o pai de um nó na árvore."""
        if root is None or root == node:
            return None

        fila = [(root, None)]

        while fila:
            atual, pai = fila.pop(0)

            if atual == node:
                return pai

            if atual.esquerdo:
                fila.append((atual.esquerdo, atual))
            if atual.direito:
                fila.append((atual.direito, atual))

        return None


    def extract_max(self):
        if not self.root:
            return None

        max_value = self.root.chave
        last_node = self._get_last_node()

        if last_node == self.root:
            self.root = None
        else:
            self.root.chave = last_node.chave
            self._remover_last_node()
            self._bubble_down(self.root)

        return max_value

    def _get_last_node(self):
        """Encontrar o último nó da árvore (para max-heap)."""
        if not self.root:
            return None

        fila = [self.root]
        last_node = self.root

        while fila:
            node = fila.pop(0)
            last_node = node

            if node.esquerdo:
                fila.append(node.esquerdo)
            if node.direito:
                fila.append(node.direito)

        return last_node

    def _remover_last_node(self):
        """Remover o último nó da árvore (para max-heap)."""
        if not self.root:
            return

        fila = [(self.root, None)]
        last_node = None
        last_parent = None

        while fila:
            node, parent = fila.pop(0)
            last_node = node
            last_parent = parent

            if node.esquerdo:
                fila.append((node.esquerdo, node))
            if node.direito:
                fila.append((node.direito, node))

        if last_parent:
            if last_parent.esquerdo == last_node:
                last_parent.esquerdo = None
            else:
                last_parent.direito = None
        else:
            self.root = None

    def _bubble_down(self, node):
        while node:
            largest = node
            if node.esquerdo and node.esquerdo.chave > largest.chave:
                largest = node.esquerdo
            if node.direito and node.direito.chave > largest.chave:
                largest = node.direito
            if largest == node:
                break
            node.chave, largest.chave = largest.chave, node.chave
            node = largest

    def insert_bst(self, chave):
        def _insert(no, chave):
            if not no:
                return No(chave)
            if chave < no.chave:
                no.esquerdo = _insert(no.esquerdo, chave)
            else:
                no.direito = _insert(no.direito, chave)
            return no
        self.root = _insert(self.root, chave)

    def remove_bst(self, chave):
        def _remover(no, chave):
            if not no:
                return no
            if chave < no.chave:
                no.esquerdo = _remover(no.esquerdo, chave)
            elif chave > no.chave:
                no.direito = _remover(no.direito, chave)
            else:
                if not no.esquerdo:
                    return no.direito
                elif not no.direito:
                    return no.esquerdo
                min_larger_no = self._get_min(no.direito)
                no.chave = min_larger_no.chave
                no.direito = _remover(no.direito, min_larger_no.chave)
            return no
        self.root = _remover(self.root, chave)

    def _get_min(self, no):
        while no.esquerdo:
            no = no.esquerdo
        return no

    def pre_ordem(self, no, resultado=None):
        if resultado is None:
            resultado = []
        if no:
            resultado.append(no.chave)
            self.pre_ordem(no.esquerdo, resultado)
            self.pre_ordem(no.direito, resultado)
        return resultado

    def em_ordem(self, no, resultado=None):
        if resultado is None:
            resultado = []
        if no:
            self.em_ordem(no.esquerdo, resultado)
            resultado.append(no.chave)
            self.em_ordem(no.direito, resultado)
        return resultado

    def pos_ordem(self, no, resultado=None):
        if resultado is None:
            resultado = []
        if no:
            self.pos_ordem(no.esquerdo, resultado)
            self.pos_ordem(no.direito, resultado)
            resultado.append(no.chave)
        return resultado

    def is_max_heap(self, no=None):
        if no is None:
            no = self.root
        if not no:
            return True

        fila = [no]

        while fila:
            atual = fila.pop(0)

            # Verifica a propriedade de max-heap para o filho esquerdo
            if atual.esquerdo:
                if atual.esquerdo.chave > atual.chave:
                    return False
                fila.append(atual.esquerdo)

            # Verifica a propriedade de max-heap para o filho direito
            if atual.direito:
                if atual.direito.chave > atual.chave:
                    return False
                fila.append(atual.direito)

        return True

    def eh_arv_bina(self, no=None, min_chave=float('-inf'), max_chave=float('inf')):
        if no is None:
            no = self.root
        if not no:
            return True
        if not (min_chave < no.chave < max_chave):
            return False
        return (self.eh_arv_bina(no.esquerdo, min_chave, no.chave) and
                self.eh_arv_bina(no.direito, no.chave, max_chave))
```

**TESTE**

```python
arvore = Arvore()

# Teste da inserção na árvore binária de busca
arvore.insert_bst(10)
arvore.insert_bst(5)
arvore.insert_bst(15)
arvore.insert_bst(2)
print("Em ordem (BST):", arvore.em_ordem(arvore.root))

# Teste de remoção na árvore binária de busca
arvore.remove_bst(5)
print("Em ordem após remoção (BST):", arvore.em_ordem(arvore.root))

# Teste da árvore max-heap
arvore.insert_max_heap(10)
arvore.insert_max_heap(15)
arvore.insert_max_heap(20)
print("Vetor (Max-Heap):", arvore.max_heap_to_vector())
print("Elemento de máxima prioridade:", arvore.get_max())

# Extração do máximo
arvore.extract_max()
print("Vetor após extrair o máximo (Max-Heap):", arvore.max_heap_to_vector())
```

Teste: Inserção e Extração em Ordem Decrescente

```python
# Criando a árvore max-heap
arvore = Arvore()

# Inserindo elementos em ordem decrescente
elementos = [100, 90, 80, 70, 60, 50, 40, 30, 20, 10]
for elem in elementos:
    arvore.insert_max_heap(elem)

# Imprimindo a árvore como vetor
print("Max-Heap após inserções (ordem decrescente):", arvore.max_heap_to_vector())

# Extraindo o máximo várias vezes para verificar a propriedade max-heap
while arvore.root:
    maximo = arvore.extract_max()
    print(f"Elemento extraído: {maximo}")
    print("Max-Heap após extração:", arvore.max_heap_to_vector())
```

Teste: Remover Nó em uma Árvore Binária de Busca (BST)

```python
# Criando a árvore binária de busca
arvore_bst = Arvore()

# Inserindo elementos na árvore binária de busca
elementos_inserir = [50, 30, 70, 20, 40, 60, 80]
for elem in elementos_inserir:
    arvore_bst.insert_bst(elem)

# Imprimindo a árvore antes da remoção
print("Árvore binária de busca (BST) antes da remoção:")
print("In-Order:", arvore_bst.em_ordem(arvore_bst.root))

# Removendo um nó folha (20)
arvore_bst.remove_bst(20)
print("\nÁrvore binária de busca (BST) após remover nó 20:")
print("In-Order:", arvore_bst.em_ordem(arvore_bst.root))

# Removendo um nó com um filho (40)
arvore_bst.remove_bst(40)
print("\nÁrvore binária de busca (BST) após remover nó 40:")
print("In-Order:", arvore_bst.em_ordem(arvore_bst.root))

# Removendo um nó com dois filhos (30, que será substituído pelo menor nó da subárvore direita)
arvore_bst.remove_bst(30)
print("\nÁrvore binária de busca (BST) após remover nó 30:")
print("In-Order:", arvore_bst.em_ordem(arvore_bst.root))
```
